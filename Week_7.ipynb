```python
import cv2
import numpy as np
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

# Baca citra dan ubah ke RGB
filename = ('/content/drive/MyDrive/PCVK 2026/lily.jpg')
img = cv2.imread(filename)
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Ubah citra menjadi array 2D (piksel)
Z = img.reshape((-1, 3))
Z = np.float32(Z)

# Tentukan jumlah cluster
k = 3
kmeans = KMeans(n_clusters=k, random_state=42)
kmeans.fit(Z)

# Dapatkan label dan pusat cluster
labels = kmeans.labels_
centers = np.uint8(kmeans.cluster_centers_)

# Bentuk citra hasil segmentasi
segmented = centers[labels.flatten()]
segmented_img = segmented.reshape(img.shape)

# Tampilkan hasil
plt.figure(figsize=(10,4))
plt.subplot(1,2,1)
plt.imshow(img)
plt.title('Citra Asli')
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(segmented_img)
plt.title('Segmentasi K-Means (k=3)')
plt.axis('off')

plt.show()
```

```python
#4
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

# 1. Baca citra Sudoku dan ubah ke Grayscale
filename = ('/content/drive/MyDrive/PCVK 2026/sudoku-original.jpg')
img = cv.imread(filename)
gray = cv.cvtColor(img, cv.COLOR_BGR2GRAY)

# 2. Parameter ketentuan soal
thresh_global = 127
block_size = 11      # Ukuran jendela 11x11 (harus ganjil)
C = 2                # Konstanta pengurang
sigma = 2            # Sigma untuk Gaussian

# 3. Global Threshold secara manual (ambang 127)
th1 = np.zeros_like(gray)
th1[gray > thresh_global] = 255
th1[gray <= thresh_global] = 0

# 4. Fungsi Manual Adaptive Mean Thresholding
def adaptive_mean_threshold(image, b_size, c_val):
    h, w = image.shape
    pad = b_size // 2
    # Padding citra menggunakan replikasi tepi (replicate padding)
    padded = np.pad(image, pad, mode='edge')
    output = np.zeros((h, w), dtype=np.uint8)

    for y in range(h):
        for x in range(w):
            # Ambil jendela lokal ukuran b_size x b_size
            window = padded[y:y+b_size, x:x+b_size]
            mean_val = np.mean(window)
            # Ambang lokal T = mean - C
            T = mean_val - c_val

            if image[y, x] > T:
                output[y, x] = 255
            else:
                output[y, x] = 0
    return output

# 5. Fungsi Manual Adaptive Gaussian Thresholding
def adaptive_gaussian_threshold(image, b_size, c_val, sig):
    h, w = image.shape
    pad = b_size // 2
    padded = np.pad(image, pad, mode='edge')
    output = np.zeros((h, w), dtype=np.uint8)

    # Buat kernel Gaussian 2D secara manual
    ax = np.arange(-pad, pad + 1)
    xx, yy = np.meshgrid(ax, ax)
    kernel = np.exp(-(xx**2 + yy**2) / (2 * sig**2))
    kernel = kernel / np.sum(kernel) # Normalisasi kernel

    for y in range(h):
        for x in range(w):
            window = padded[y:y+b_size, x:x+b_size]
            # Hitung rata-rata berbobot Gaussian
            gauss_mean = np.sum(window * kernel)
            T = gauss_mean - c_val

            if image[y, x] > T:
                output[y, x] = 255
            else:
                output[y, x] = 0
    return output

# Eksekusi fungsi manual
print("Menghitung Adaptive Mean Threshold (manual)...")
th2 = adaptive_mean_threshold(gray, block_size, C)

print("Menghitung Adaptive Gaussian Threshold (manual)...")
th3 = adaptive_gaussian_threshold(gray, block_size, C, sigma)

# 6. Plotting Hasil
titles = ['Citra Asli (Grayscale)', 'Global Thresholding Manual (v = 127)',
          'Adaptive Mean Thresholding (Manual)', 'Adaptive Gaussian Thresholding (Manual)']
citra2 = [gray, th1, th2, th3]

plt.figure(figsize = (10, 10))
for i in range(len(citra2)):
    plt.subplot(2, 2, i + 1), plt.imshow(citra2[i], 'gray')
    plt.title(titles[i])
    plt.xticks([]), plt.yticks([])

plt.show()
```

```python
#5
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

# 1. Baca citra grayscale dan lakukan Gaussian Blur
filename = ('/content/drive/MyDrive/PCVK 2026/lena_gs_lc2.jpg')
img = cv.imread(filename, 0)
blur = cv.GaussianBlur(img, (5, 5), 0)
thresh = 127

# 2. Fungsi Otsu's Thresholding Manual
def otsu(gray):
    pixel_number = gray.shape[0] * gray.shape[1]
    mean_weight = 1.0 / pixel_number
    his, bins = np.histogram(gray, np.arange(0, 257))
    final_thresh = -1
    final_value = -1
    intensity_arr = np.arange(256)

    for t in bins[1:-1]:
        pcb = np.sum(his[:t])
        pcf = np.sum(his[t:])

        # Jika salah satu kelas kosong, lewati
        if pcb == 0 or pcf == 0:
            continue

        Wb = pcb * mean_weight
        Wf = pcf * mean_weight

        mub = np.sum(intensity_arr[:t] * his[:t]) / float(pcb)
        muf = np.sum(intensity_arr[t:] * his[t:]) / float(pcf)

        # Hitung between-class variance
        value = Wb * Wf * (mub - muf) ** 2

        if value > final_value:
            final_thresh = t
            final_value = value

    final_img = gray.copy()
    print("Nilai threshold optimal Otsu:", final_thresh)
    final_img[gray > final_thresh] = 255
    final_img[gray <= final_thresh] = 0
    return final_img, final_thresh

# Eksekusi fungsi Otsu manual
otsu_biner, otsu_thresh = otsu(blur)
x = ("Otsu's Thresholding (v = ") + str(otsu_thresh) + ")"

# Global Thresholding konvensional (v = 127)
ret, th1 = cv.threshold(blur, thresh, 255, cv.THRESH_BINARY)

# 3. Plotting Hasil Perbandingan
titles = ['Citra Asli', 'Global Thresholding (Thresh = ' + str(thresh) + ')', x]
citra3 = [blur, th1, otsu_biner]

plt.figure(figsize = (20, 15))
for i in range(len(citra3)):
    plt.subplot(1, 3, i + 1), plt.imshow(citra3[i], 'gray')
    plt.title(titles[i])
    plt.xticks([]), plt.yticks([])

plt.show()
```

```python
#6
import matplotlib.pyplot as plt
import numpy as np
from skimage import data
from skimage.segmentation import flood

# 1. Memuat citra data koin dari skimage
img = data.coins()  # Citra grayscale ukuran 303 x 384

# 2. Tentukan 3 titik seed (koordinat baris, kolom / y, x) sesuai ketentuan:
#    - Seed 1: Di dalam salah satu koin
#    - Seed 2: Di latar belakang (background)
#    - Seed 3: Tepat di garis tepi koin
seeds = [
    (36, 43),    # Seed 1: Di dalam koin bagian atas
    (15, 15),    # Seed 2: Di latar belakang
    (115, 145)   # Seed 3: Di tepi koin
]

# 3. Variasi nilai toleransi (T) yang diuji
thresholds = [5, 20, 60]

# 4. Membuat visualisasi grid 3x3 (Baris = Seed, Kolom = Nilai T)
fig, axes = plt.subplots(3, 3, figsize=(12, 12))

for i, seed in enumerate(seeds):
    for j, t in enumerate(thresholds):
        # Jalankan algoritma Region Growing / Flood Fill
        mask = flood(img, seed_point=seed, tolerance=t)

        # Tampilkan citra asli
        axes[i, j].imshow(img, cmap='gray')

        # Overlay hasil mask region dengan colormap 'jet' dan transparansi (alpha)
        axes[i, j].imshow(mask, cmap='jet', alpha=0.4)

        # Tandai posisi titik seed dengan bulatan merah kecil pada kolom pertama
        if j == 0:
            axes[i, j].plot(seed[1], seed[0], 'ro', markersize=5)

        axes[i, j].set_title(f"Seed {i+1} (y={seed[0]}, x={seed[1]})\nT = {t}")
        axes[i, j].axis('off')

plt.tight_layout()
plt.show()
```

```python
#6
import matplotlib.pyplot as plt
from skimage import data
from skimage.segmentation import flood

# 1. Muat citra grayscale data koin dari skimage
gray = data.coins()

# 2. Tentukan koordinat titik seed (baris, kolom) sesuai contoh pada modul
seed = (36, 43)

# 3. Definisikan variasi nilai toleransi (T) yang diminta
thresholds = [5, 20, 60]

# 4. Buat visualisasi 4 panel berjajar mendatar (Citra Asli + Seed & 3 Variasi T)
fig, axes = plt.subplots(1, 4, figsize=(16, 4))

# Panel 1: Citra Asli dengan titik seed merah
axes[0].imshow(gray, cmap='gray')
axes[0].plot(seed[1], seed[0], 'ro', markersize=6)
axes[0].set_title("Citra asli + seed")
axes[0].axis('off')

# Panel 2, 3, 4: Hasil flood() dengan T = 5, 20, dan 60
for j, t in enumerate(thresholds):
    # Jalankan fungsi flood() dari skimage.segmentation
    region = flood(gray, seed_point=seed, tolerance=t, connectivity=1)

    # Hitung jumlah piksel yang masuk ke dalam mask region (untuk info teks)
    pixel_count = region.sum()

    axes[j+1].imshow(gray, cmap='gray')
    axes[j+1].imshow(region, cmap='jet', alpha=0.4)
    axes[j+1].set_title(f"Region Growing, T = {t}\n({pixel_count} piksel)")
    axes[j+1].axis('off')

plt.tight_layout()
plt.show()
```

```python
#7
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

# 1. Baca citra KTP dari Google Drive / path lokal
filename = ('/content/drive/MyDrive/PCVK 2026/ktm.png')
img = cv.imread(filename)

# Jika citra berhasil dibaca, konversi dari BGR ke RGB (untuk plotting matplotlib)
# dan konversi ke HSV untuk proses segmentasi warna
if img is not None:
    img_rgb = cv.cvtColor(img, cv.COLOR_BGR2RGB)
    hsv = cv.cvtColor(img, cv.COLOR_BGR2HSV)

    # Ambil komponen kanal warna H (Hue), S (Saturation), dan V (Value)
    H = hsv[:, :, 0]
    S = hsv[:, :, 1]
    V = hsv[:, :, 2]

    # 2. Terapkan kondisi rentang warna biru sesuai petunjuk modul
    kelompok_biru = np.where((H >= 85) & (H <= 130) & (S >= 25) & (V >= 40))

    # 3. Buat citra kosong (hitam) untuk menampilkan hanya warna biru
    h_img, w_img, _ = img.shape
    hanya_biru = np.zeros((h_img, w_img, 3), dtype=np.uint8)

    # Salin piksel warna asli pada koordinat yang memenuhi kelompok_biru
    hanya_biru[kelompok_biru] = img_rgb[kelompok_biru]

    # 4. Plotting hasil perbandingan (Citra Asli, Segmentasi K-Means / Mask, Hasil Warna Biru Saja)
    plt.figure(figsize=(15, 5))

    plt.subplot(1, 3, 1)
    plt.imshow(img_rgb)
    plt.title('Citra Asli')
    plt.axis('off')

    # Visualisasi mask area biru biner (hitam-putih)
    mask_biner = np.zeros((h_img, w_img), dtype=np.uint8)
    mask_biner[kelompok_biru] = 255

    plt.subplot(1, 3, 2)
    plt.imshow(mask_biner, cmap='gray')
    plt.title('Mask Area Biru')
    plt.axis('off')

    plt.subplot(1, 3, 3)
    plt.imshow(hanya_biru)
    plt.title('Hanya Warna Biru Saja')
    plt.axis('off')

    plt.show()
else:
    print("Gagal memuat gambar. Pastikan path direktori file ktp_fulan.jpg sudah benar.")
```

```python
#Soal 2
# Daftar path citra asli
image_paths = [
    '/content/drive/MyDrive/PCVK 2026/ktp_karim.jpeg',
    '/content/drive/MyDrive/PCVK 2026/ktp_abit.jpeg',
    '/content/drive/MyDrive/PCVK 2026/ktp_yosan.jpeg'
]

names = ['Karim', 'Abit', 'Yosan']

# Pengaturan konsisten: bahasa Indonesia dan psm 6
custom_config = r'--oem 3 --psm 6'

for i, path in enumerate(image_paths):
    # Membaca citra
    img_color = cv.imread(path)

    if img_color is not None:
        # Konversi BGR ke RGB (Tesseract membaca format RGB lebih akurat)
        img_rgb = cv.cvtColor(img_color, cv.COLOR_BGR2RGB)

        print(f"==================================================")
        print(f" HASIL PEMBACAAN TEKS OCR - KTP {names[i].upper()} (CITRA ASLI)")
        print(f"==================================================")

        # Ekstraksi teks pada citra berwarna asli
        text_color = pytesseract.image_to_string(img_rgb, lang='ind', config=custom_config)
        print(text_color if text_color.strip() else "(Tidak ada teks terbaca)")
        print("\n")
    else:
        print(f"Peringatan: Gambar tidak ditemukan di path '{path}'")
```

```python
#Soal 3
# Daftar path citra
image_paths = [
    '/content/drive/MyDrive/PCVK 2026/ktp_karim.jpeg',
    '/content/drive/MyDrive/PCVK 2026/ktp_abit.jpeg',
    '/content/drive/MyDrive/PCVK 2026/ktp_yosan.jpeg'
]

names = ['Karim', 'Abit', 'Yosan']

# Pengaturan konsisten: bahasa Indonesia dan psm 6
custom_config = r'--oem 3 --psm 6'
conf_threshold = 60  # Batas confidence > 60

def draw_text_boxes(image, is_gray=False):
    """
    Fungsi untuk menjalankan OCR, menggambar kotak hijau,
    dan menghitung rata-rata confidence.
    """
    if is_gray:
        img_for_ocr = image
        img_canvas = cv.cvtColor(image, cv.COLOR_GRAY2RGB)
    else:
        img_for_ocr = cv.cvtColor(image, cv.COLOR_BGR2RGB)
        img_canvas = img_for_ocr.copy()

    d = pytesseract.image_to_data(img_for_ocr, lang='ind', config=custom_config, output_type=Output.DICT)
    n_boxes = len(d['text'])

    total_conf = 0
    valid_words = 0

    for i in range(n_boxes):
        conf = int(d['conf'][i])

        if conf > conf_threshold:
            (x, y, w, h) = (d['left'][i], d['top'][i], d['width'][i], d['height'][i])
            cv.rectangle(img_canvas, (x, y), (x + w, y + h), (0, 255, 0), 2)

            # Tambahkan perhitungan rata-rata
            total_conf += conf
            valid_words += 1

    # Hitung rata-rata confidence
    avg_conf = total_conf / valid_words if valid_words > 0 else 0
    return img_canvas, avg_conf

plt.figure(figsize=(20, 15))

for i, path in enumerate(image_paths):
    img_color = cv.imread(path)

    if img_color is not None:
        img_gray = cv.cvtColor(img_color, cv.COLOR_BGR2GRAY)
        ret, img_otsu = cv.threshold(img_gray, 0, 255, cv.THRESH_BINARY + cv.THRESH_OTSU)
        img_adaptive = cv.adaptiveThreshold(img_gray, 255, cv.ADAPTIVE_THRESH_GAUSSIAN_C, cv.THRESH_BINARY, 11, 2)

        # Ambil gambar berkotak sekaligus nilai rata-rata confidence-nya
        box_color, conf_color = draw_text_boxes(img_color, is_gray=False)
        box_otsu, conf_otsu = draw_text_boxes(img_otsu, is_gray=True)
        box_adaptive, conf_adaptive = draw_text_boxes(img_adaptive, is_gray=True)

        # Cetak nilai rata-rata confidence ke layar untuk diisi ke tabel
        print(f"--- Rata-rata Confidence KTP {names[i]} ---")
        print(f"Berwarna Asli : {conf_color:.2f}%")
        print(f"Otsu          : {conf_otsu:.2f}%")
        print(f"Adaptive      : {conf_adaptive:.2f}%\n")

        plt.subplot(3, 3, i*3 + 1)
        plt.imshow(box_color)
        plt.title(f'Area Teks - Berwarna ({names[i]})\nConf: {conf_color:.1f}%')
        plt.axis('off')

        plt.subplot(3, 3, i*3 + 2)
        plt.imshow(box_otsu)
        plt.title(f'Area Teks - Otsu ({names[i]})\nConf: {conf_otsu:.1f}%')
        plt.axis('off')

        plt.subplot(3, 3, i*3 + 3)
        plt.imshow(box_adaptive)
        plt.title(f'Area Teks - Adaptive ({names[i]})\nConf: {conf_adaptive:.1f}%')
        plt.axis('off')
    else:
        print(f"Peringatan: Gambar tidak ditemukan di path '{path}'")

plt.tight_layout()
plt.show()
```
